```python
import matplotlib.pyplot as plt
import torch
```

```python
# Load the checkpoint
checkpoint = torch.load("models/DualStreamNetwork_Adam_lr0.0001_dropout0.5_epochs20_batch64_nframes1_nflow9_test0.5333_withourreset.pth", map_location="cpu")  # adjust path/filename as needed

# Access train and validation accuracies
train_accuracies = checkpoint['train_accuracies']
val_accuracies = checkpoint['val_accuracies']
```

```python
plt.plot(train_accuracies, label='Train Accuracy')
plt.plot(val_accuracies, label='Validation Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.title('Training and Validation Accuracy over Epochs')
plt.legend()
```
